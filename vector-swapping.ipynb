```python
class TokenIdResolver:
    """
    Resolves word(s) to single vocabulary token IDs, raising loud errors on
    multi-token words.
    """

    def __init__(self, tokenizer):
        self.tok = tokenizer

    def resolve(self, word: str) -> int:
        """Single word -> single token id. Raises if it's not one token."""
        ids = self.tok.encode(word, add_special_tokens=False)
        if len(ids) != 1:
            decoded = [self.tok.decode([i]) for i in ids]
            raise ValueError(
                f"{word!r} tokenizes to {len(ids)} tokens: {ids} {decoded} — "
                f"not a single token. Pick a different word or handle."
            )
        return ids[0]

    def resolve_pair(self, source_word: str, target_word: str) -> tuple[int, int]:
        """Convenience for the swap use case: resolves both at once,
        reporting BOTH failures together if either (or both) fail,
        rather than stopping at the first one."""
        errors = []
        source_id = target_id = None

        try:
            source_id = self.resolve(source_word)
        except ValueError as e:
            errors.append(str(e))

        try:
            target_id = self.resolve(target_word)
        except ValueError as e:
            errors.append(str(e))

        if errors:
            raise ValueError("Failed to resolve pair:\n" + "\n".join(errors))

        return source_id, target_id

    def filter_valid(self, words: list[str]) -> tuple[list[str], list[str]]:
        """Splits a word list into (single-token words, multi-token words) —
        useful for auditing a whole stimulus list up front, per the
        single-token-filtering strategy from earlier."""
        valid, invalid = [], []
        for w in words:
            try:
                self.resolve(w)
                valid.append(w)
            except ValueError:
                invalid.append(w)
        return valid, invalid
```

```python
class TwoPassLensSwap:
    """
    We swap tokens by first running one clean pass to find the coefficients of each vector in the activation, and then a second pass where we swap these coeffecients.
    This setup allows us to theoretically swap solely the roles each token plays in the activation (h) at that layer and leave the rest of h unaltered.
    
    Pass 1 (clean):    for each layer l in band: cache c_l = V_l⁺ h_l
    Pass 2 (swapped):  for each layer l in band: force h_l's coordinates
                        toward flip(c_l), the CACHED clean value.
    """

    def __init__(self, hf_model, lens, W_U, layers, source_id, target_id,
                 positions=None, alpha=1.0):
        self.hf_model = hf_model
        self.layers = list(layers)
        self.positions = positions
        self.alpha = alpha

        self.V = {}
        self.V_pinv = {}
        for layer in self.layers:
            J_l = lens.jacobians[layer].to(device=W_U.device, dtype=torch.float32)
            w_s = W_U[source_id].to(torch.float32)
            w_t = W_U[target_id].to(torch.float32)
            v_s = w_s @ J_l
            v_t = w_t @ J_l
            V = torch.stack([v_s, v_t], dim=1)
            self.V[layer] = V
            self.V_pinv[layer] = torch.linalg.pinv(V)
        self.cached_c = {}

    #Pass 1:
    def _make_read_hook(self, layer):
        V_pinv = self.V_pinv[layer]

        def hook(module, input, output):
            h = output[0] if isinstance(output, tuple) else output
            pos = self.positions if self.positions is not None else slice(None)
            h_sel = h[:, pos, :].float()
            c = h_sel @ V_pinv.T
            self.cached_c[layer] = c.detach()
            return output
        return hook

    def run_clean_pass(self, input_ids):
        handles = []
        try:
            for layer in self.layers:
                h = self.hf_model.model.layers[layer].register_forward_hook(
                    self._make_read_hook(layer)
                )
                handles.append(h)
            with torch.inference_mode():
                out = self.hf_model(input_ids)
        finally:
            for h in handles:
                h.remove()
        return out

    #Pass 2:
    def _make_swap_hook(self, layer):
        V = self.V[layer]
        V_pinv = self.V_pinv[layer]
        
        # Fires only on the first forward through this layer.
        # Subsequent forwards (decode steps) pass through untouched.
        state = {"applied": False}

        def hook(module, input, output):
            if state["applied"]:
                return output
            state["applied"] = True

            h = output[0] if isinstance(output, tuple) else output
            pos = self.positions if self.positions is not None else slice(None)
            h_sel = h[:, pos, :]
            h_sel_f32 = h_sel.float()

            c_current = h_sel_f32 @ V_pinv.T
            c_clean_swapped = self.cached_c[layer].flip(dims=[-1])
            c_new = c_current + self.alpha * (c_clean_swapped - c_current)

            delta = ((c_new - c_current) @ V.T).to(h.dtype)
            h = h.clone()
            h[:, pos, :] = h_sel + delta

            if isinstance(output, tuple):
                return (h,) + output[1:]
            return h
        return hook

    def run_swapped_pass(self, input_ids):
        if not self.cached_c:
            raise RuntimeError("run_clean_pass must be called before run_swapped_pass")
        handles = []
        try:
            for layer in self.layers:
                h = self.hf_model.model.layers[layer].register_forward_hook(
                    self._make_swap_hook(layer)
                )
                handles.append(h)
            with torch.inference_mode():
                out = self.hf_model(input_ids)
        finally:
            for h in handles:
                h.remove()
        return out

    def run(self, input_ids):
        clean_out = self.run_clean_pass(input_ids)
        swapped_out = self.run_swapped_pass(input_ids)
        return clean_out, swapped_out

    #Generation
    @torch.inference_mode()
    def generate(self, input_ids, max_new_tokens=50, swapped=True, **gen_kwargs):
        """
        Generate up to `max_new_tokens` continuations.
        If swapped=False: plain clean generation (no hooks).
        If swapped=True : pass 1 caches clean c's; pass 2 registers swap hooks
                          that fire ONLY on the prefill forward, then disable
                          themselves — so generation itself is unperturbed.
        """
        if swapped and not self.cached_c:
            self.run_clean_pass(input_ids)

        handles = []
        if swapped:
            for layer in self.layers:
                h = self.hf_model.model.layers[layer].register_forward_hook(
                    self._make_swap_hook(layer)
                )
                handles.append(h)

        gen_kwargs.setdefault("do_sample", False)
        gen_kwargs.setdefault("pad_token_id",
                              self.hf_model.config.eos_token_id)
        try:
            out = self.hf_model.generate(
                input_ids,
                max_new_tokens=max_new_tokens,
                **gen_kwargs,
            )
        finally:
            for h in handles:
                h.remove()
        return out

    @torch.inference_mode()
    def generate_both(self, input_ids, max_new_tokens=50, **gen_kwargs):
        """Returns (clean_ids, swapped_ids) — only prompt activations differ."""
        clean_ids = self.generate(input_ids, max_new_tokens,
                                  swapped=False, **gen_kwargs)
        swapped_ids = self.generate(input_ids, max_new_tokens,
                                    swapped=True, **gen_kwargs)
        return clean_ids, swapped_ids
```

```python
#Helper Functions

def _get_final_norm(hf_model):
    if hasattr(hf_model, "model") and hasattr(hf_model.model, "norm"):
        return hf_model.model.norm
    if hasattr(hf_model, "transformer") and hasattr(hf_model.transformer, "ln_f"):
        return hf_model.transformer.ln_f
    raise AttributeError("Could not find final norm on model")


def _jlens_logits(h_l, layer, lens, final_norm, W_U, transpose=False):
    """
    h_l: residual at band layer `layer`.
    Storage convention (Anthropic repo): J_l is Transposed in storage to match h, so transport
    is  h_l @ J_l.T. Perhaps some lenses store it another way,
    pass transpose=True to use  h_l @ J_l.
    """
    J_l = lens.jacobians[layer].to(device=h_l.device, dtype=torch.float32)
    if transpose:
        h_final = h_l.float() @ J_l
    else:
        h_final = h_l.float() @ J_l.T

    h_final = h_final.to(final_norm.weight.dtype)
    h_normed = final_norm(h_final)
    logits = h_normed @ W_U.T.to(h_normed.dtype)
    return logits.float()


def _jlens_top_k(h_l, layer, lens, final_norm, W_U, tokenizer,
                 k=10, positions=None, transpose=False):
    logits = _jlens_logits(h_l, layer, lens, final_norm, W_U, transpose)
    probs = torch.softmax(logits[0], dim=-1)          # [S, vocab]
    S = probs.shape[0]
    pos_range = positions if positions is not None else range(S)

    out = []
    for pos in pos_range:
        tp, ti = probs[pos].topk(k)
        toks = [tokenizer.decode([t]) for t in ti.tolist()]
        out.append(list(zip(toks, tp.tolist())))
    return out


def _lens_layers(lens):
    """Return the set of layer indices the lens has Jacobians for."""
    j = lens.jacobians
    return set(j.keys()) if hasattr(j, "keys") else set(range(len(j)))


def _run_and_capture(hf_model, layers, input_ids, capture_store,
                     swap_hooks=None, read_hooks=None, **model_kwargs):
    """
    Run one forward pass, capturing post-block residuals at each layer in
    `layers` into capture_store[layer].

    swap_hooks: optional list of (layer, hook_fn) — applied first, so the
                captured residual is the post-swap one.
    read_hooks: optional list of (layer, hook_fn) — applied before capture
                hooks so they see the pre-swap residual (used to populate
                swapper.cached_c during the clean pass).
    """
    capture_store.clear()
    handles = []

    if swap_hooks:
        for layer, hook in swap_hooks:
            handles.append(
                hf_model.model.layers[layer].register_forward_hook(hook)
            )
    if read_hooks:
        for layer, hook in read_hooks:
            handles.append(
                hf_model.model.layers[layer].register_forward_hook(hook)
            )

    def make_capture(layer):
        def hook(module, input, output):
            h = output[0] if isinstance(output, tuple) else output
            capture_store[layer] = h.detach()
            return output
        return hook

    for layer in layers:
        handles.append(
            hf_model.model.layers[layer].register_forward_hook(make_capture(layer))
        )

    try:
        with torch.inference_mode():
            hf_model(input_ids, **model_kwargs)
    finally:
        for h in handles:
            h.remove()


#Comparison function

def compare_clean_vs_swapped(clean_out, swapped_out, tokenizer, input_ids,
                             k=10, positions=None,
                             swapper=None, lens=None, transpose=False,
                             **model_kwargs):
    """
    Prints J-lens top-k at every captured layer, clean vs swapped, per position.
    """
    if swapper is None or lens is None:
        raise ValueError(
            "compare_clean_vs_swapped needs `swapper=` and `lens=` to "
            "compute J-lens readouts. Pass the same objects you gave to "
            "TwoPassLensSwap."
        )

    hf_model   = swapper.hf_model
    layers     = swapper.layers
    final_norm = _get_final_norm(hf_model)
    W_U        = hf_model.lm_head.weight

    lens_layers = _lens_layers(lens)
    final_layer = max(lens_layers)

    # Only include layers the lens can actually read out.
    requested      = sorted(set(list(layers) + [final_layer]))
    capture_layers = [l for l in requested if l in lens_layers]

    missing = [l for l in requested if l not in lens_layers]
    if missing:
        print(f"[compare_clean_vs_swapped] skipping layers with no lens: {missing}")

    #Pass 1: clean
    read_hooks = [(l, swapper._make_read_hook(l)) for l in layers]
    clean_resid = {}
    _run_and_capture(hf_model, capture_layers, input_ids, clean_resid,
                     read_hooks=read_hooks, **model_kwargs)

    #Pass 2: swapped
    swap_hooks = [(l, swapper._make_swap_hook(l)) for l in layers]
    swapped_resid = {}
    _run_and_capture(hf_model, capture_layers, input_ids, swapped_resid,
                     swap_hooks=swap_hooks, **model_kwargs)

    #J-lens readout at each captured layer
    pos_range = (positions if positions is not None
                 else range(input_ids.shape[1]))

    for layer in capture_layers:
        clean_top = _jlens_top_k(clean_resid[layer], layer, lens,
                                 final_norm, W_U, tokenizer,
                                 k=k, positions=list(pos_range),
                                 transpose=transpose)
        swapped_top = _jlens_top_k(swapped_resid[layer], layer, lens,
                                   final_norm, W_U, tokenizer,
                                   k=k, positions=list(pos_range),
                                   transpose=transpose)

        label = f"layer {layer}" + ("  (LENS-FINAL)" if layer == final_layer else "")
        print(f"\n================= {label} =================")
        for i, pos in enumerate(pos_range):
            print(f"--- position {pos} ---")
            print(f"{'CLEAN J-LENS':<32} {'SWAPPED J-LENS':<32}")
            for j in range(k):
                c_tok, c_p = clean_top[i][j]
                s_tok, s_p = swapped_top[i][j]
                print(f"{c_tok!r:<20}{c_p:>8.4f}   {s_tok!r:<20}{s_p:>8.4f}")

    return clean_resid, swapped_resid
```
